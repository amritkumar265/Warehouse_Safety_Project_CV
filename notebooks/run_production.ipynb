```python
from pathlib import Path

PROJECT_ROOT = Path("/content/drive/MyDrive/CCTV_VLM_Project")

pages = {
    "video_analysis.py": ("🎬 Video Analysis", "Upload and analyze warehouse CCTV footage."),
    "detection_lab.py": ("🔍 Detection Lab", "YOLO, Grounding DINO and detection fusion."),
    "tracking_lab.py": ("🎯 Tracking Lab", "ByteTrack and track validation."),
    "safety_audit.py": ("⚠️ Safety Audit", "Near-miss, obstruction and PPE analysis."),
    "vlm_analysis.py": ("🧠 VLM Analysis", "Language-driven scene understanding and PPE analysis."),
    "alerts_reports.py": ("📊 Alerts & Reports", "Safety events, reports and notifications.")
}

for filename, (title, description) in pages.items():
    content = f'''import streamlit as st

st.title("{title}")
st.caption("{description}")

st.info("This module is under development.")
'''
    (PROJECT_ROOT / "pages" / filename).write_text(content)

    print("Created:", filename)
```
