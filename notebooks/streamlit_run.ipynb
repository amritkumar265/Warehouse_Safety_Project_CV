```python
import subprocess

print("1. STREAMLIT PROCESS")
subprocess.run(["bash", "-lc", "ps aux | grep '[s]treamlit'"])

print("\n2. PORT 8501 HEALTH CHECK")
subprocess.run([
    "bash", "-lc",
    "curl -v --max-time 5 http://127.0.0.1:8501/_stcore/health"
])

print("\n3. PORT 8502 HEALTH CHECK")
subprocess.run([
    "bash", "-lc",
    "curl -v --max-time 5 http://127.0.0.1:8502/_stcore/health"
])
```

```python
import subprocess

commands = {
    "Streamlit processes": "ps aux | grep '[s]treamlit'",
    "Port 8501": "curl -sS --max-time 5 http://127.0.0.1:8501/_stcore/health",
    "Port 8502": "curl -sS --max-time 5 http://127.0.0.1:8502/_stcore/health"
}

for name, command in commands.items():
    print(f"\n{'='*20} {name} {'='*20}")

    result = subprocess.run(
        command,
        shell=True,
        capture_output=True,
        text=True
    )

    print("OUTPUT:", result.stdout or "No response")
    print("ERROR:", result.stderr or "None")
    print("EXIT CODE:", result.returncode)
```
